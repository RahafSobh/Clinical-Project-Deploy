```python
import sys
import joblib
PROJECT_PATH = "/content/predictor2/predictor"
sys.path.append(PROJECT_PATH)

icu_los_triage = joblib.load(PROJECT_PATH + "/outputs/icu_los_triage.joblib")
mortality_triage = joblib.load(PROJECT_PATH + "/outputs/mortality_triage.joblib")
print("Models loaded successfully!")
print(type(icu_los_triage))
print(type(mortality_triage))
```

```python
from flask import Flask, request, jsonify
from flask_cors import CORS
import pandas as pd
import threading

app = Flask(__name__)
CORS(app)

def build_patient_row(data):
    diagnosis_codes = data.get("diagnosis_codes", [])

    if isinstance(diagnosis_codes, str):
        diagnosis_codes = [diagnosis_codes]

    row = {
        "SUBJECT_ID": data.get("subject_id", 0),
        "HADM_ID": data.get("hadm_id", 0),

        "age": float(data.get("age", 60)),
        "ed_los_hours": float(data.get("ed_los_hours", 0)),
        "n_diagnoses": len(diagnosis_codes),
        "is_newborn": int(data.get("is_newborn", 0)),

        "gender": data.get("gender", "UNKNOWN"),
        "admission_type": data.get("admission_type", "EMERGENCY"),
        "admission_location": data.get("admission_location", "UNKNOWN"),
        "insurance": data.get("insurance", "UNKNOWN"),
        "marital_status": data.get("marital_status", "UNKNOWN"),
        "ethnicity": data.get("ethnicity", "UNKNOWN"),
        "religion": data.get("religion", "UNKNOWN"),
        "language": data.get("language", "UNKNOWN"),
        "came_via_ed": data.get("came_via_ed", "UNKNOWN"),

        "first_careunit": data.get("first_careunit", "MICU"),
        "adm_diagnosis": data.get("adm_diagnosis", data.get("main_diagnosis", "UNKNOWN")),
        "icd9_str": " ".join(diagnosis_codes),

        "mortality": 0,
        "icu_los_days": 0,
    }

    return pd.DataFrame([row])


def risk_level(probability):
    if probability < 0.10:
        return "Low"
    elif probability < 0.30:
        return "Medium"
    else:
        return "High"


def support_note(los_days, risk):
    if risk == "High" or los_days >= 7:
        return "Patient may require higher ICU resource planning and closer monitoring."
    elif risk == "Medium" or los_days >= 3:
        return "Patient may require medium ICU resource planning."
    else:
        return "Patient appears to require lower ICU resource planning based on the model output."


@app.route("/")
def home():
    return jsonify({
        "status": "API is running",
        "message": "ICU prediction backend is ready"
    })


@app.route("/predict", methods=["POST"])
def predict():
    try:
        data = request.get_json()
        X = build_patient_row(data)

        predicted_los = float(icu_los_triage.predict(X)[0])
        mortality_probability = float(mortality_triage.predict_proba(X)[0][1])
        risk = risk_level(mortality_probability)

        return jsonify({
            "predicted_icu_los_days": round(predicted_los, 2),
            "mortality_probability": round(mortality_probability, 3),
            "mortality_risk_level": risk,
            "decision_support_note": support_note(predicted_los, risk)
        })

    except Exception as e:
        return jsonify({
            "error": str(e)
        }), 500


def run_app():
    app.run(host="0.0.0.0", port=5000)

threading.Thread(target=run_app).start()
```
